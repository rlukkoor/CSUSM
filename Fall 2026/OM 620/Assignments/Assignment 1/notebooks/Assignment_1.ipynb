# Question 4 — Extracting The Country Values
To apply broader breakdowns of your transactional dataset, it will be helpful to add a country column. Recall that the metadata indicates the country information is somewhat captured in the site column. Create a new country column by using the manufacturing_site information. To do so, you will need the helper function str.startswith() (here is a link to the official documentation), which is defined for a pandas Series object. As the documentation states, str.startswith() tests if the start of each string element matches a given pattern.

````python
unique_values = transaction_data_nan_cleaned["manufacturing_site"].unique()
unique_values
````

````python
transaction_data_nan_cleaned['country'] = None

transaction_data_nan_cleaned.loc[transaction_data_nan_cleaned['manufacturing_site'].str.startswith(
    'PL', na=False), 'country'] = 'PL'
transaction_data_nan_cleaned.loc[transaction_data_nan_cleaned['manufacturing_site'].str.startswith(
    'JP', na=False), 'country'] = 'JP'
transaction_data_nan_cleaned.loc[transaction_data_nan_cleaned['manufacturing_site'].str.startswith(
    'US', na=False), 'country'] = 'US'
transaction_data_nan_cleaned.loc[transaction_data_nan_cleaned['manufacturing_site'].str.startswith(
    'CA', na=False), 'country'] = 'CA'

transaction_data_nan_cleaned
````

We started by using the ```unique()``` function to see how many different countries are listed under the manufacturing_site column. From this we got 4 countries: JP, CA, US, and PL.
Using the ```str.startswith()``` function, we took the first two characters from the manufacturing_site column and placed them in a new column called country.
